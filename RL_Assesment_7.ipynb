```python
class GridWorldEnv:
    def __init__(self):
        # a. State Space
        # S = {S0, S1, S3, S4, G} (X is an obstacle and not a valid state)
        self.states = ["S0", "S1", "S3", "S4", "G"]

        # b. Action Space
        self.actions = ["Up", "Down", "Left", "Right"]

        # d. Discount Factor (gamma)
        self.gamma = 0.9

        # Grid positions for reference: (row, col)
        self.state_positions = {
            "S0": (0, 0),
            "S1": (0, 1),
            "G": (0, 2),
            "S3": (1, 1),
            "S4": (1, 2)
        }

        # c. & e. Transition & Reward Mapping
        # Table format: (current_state, action) -> (next_state, reward, is_terminal)
        self.transitions = {
            ("S0", "Up"): ("S0", -1.0, False),
            ("S0", "Down"): ("S0", -5.0, True),    # Moves into Obstacle X -> -5 reward & terminal
            ("S0", "Left"): ("S0", -1.0, False),
            ("S0", "Right"): ("S1", -1.0, False),

            ("S1", "Up"): ("S1", -1.0, False),
            ("S1", "Down"): ("S3", -1.0, False),
            ("S1", "Left"): ("S0", -1.0, False),
            ("S1", "Right"): ("G", 10.0, True),    # Reaches Goal G -> +10 reward & terminal

            ("S3", "Up"): ("S1", -1.0, False),
            ("S3", "Down"): ("S3", -1.0, False),
            ("S3", "Left"): ("S3", -5.0, True),    # Moves into Obstacle X -> -5 reward & terminal
            ("S3", "Right"): ("S4", -1.0, False),

            ("S4", "Up"): ("G", 10.0, True),       # Reaches Goal G -> +10 reward & terminal
            ("S4", "Down"): ("S4", -1.0, False),
            ("S4", "Left"): ("S3", -1.0, False),
            ("S4", "Right"): ("S4", -1.0, False),
        }

    def get_state_space(self):
        """Returns state space."""
        return self.states

    def get_action_space(self):
        """Returns action space."""
        return self.actions

    def get_discount_factor(self):
        """Returns discount factor gamma."""
        return self.gamma

    def transition_function(self, state, action):
        """
        e. & f. Estimate next state and reward based on current state and action.
        Returns: (next_state, reward, done)
        """
        if state == "G":
            print(f"State {state} is terminal. No transitions out.")
            return "G", 0.0, True

        if (state, action) in self.transitions:
            return self.transitions[(state, action)]
        else:
            raise ValueError(f"Invalid state '{state}' or action '{action}' provided.")
```

```python
def calculate_returns(rewards, gamma):
    """Assignment 1: Calculates undiscounted and discounted returns."""
    # Undiscounted return = sum of all rewards
    undiscounted_return = sum(rewards)

    # Discounted return = sum( (gamma^t) * reward_t )
    discounted_return = 0.0
    for t, reward in enumerate(rewards):
        discounted_return += (gamma ** t) * reward

    return undiscounted_return, discounted_return


def run_assignment_1():
    print("=" * 50)
    print("ASSIGNMENT 1: Calculate Undiscounted and Discounted Returns")
    print("=" * 50)

    try:
        num_steps = int(input("Enter total number of time steps/episodes: "))
        rewards = []
        for i in range(num_steps):
            r = float(input(f"Enter reward for step {i + 1}: "))
            rewards.append(r)

        gamma = float(input("Enter discount factor (gamma, e.g., 0.9): "))

        undiscounted, discounted = calculate_returns(rewards, gamma)

        print("\n--- RESULTS ---")
        print(f"Entered Rewards Sequence: {rewards}")
        print(f"Discount Factor (gamma): {gamma}")
        print(f"Undiscounted Return: {undiscounted:.4f}")
        print(f"Discounted Return: {discounted:.4f}\n")

    except ValueError:
        print("Invalid input. Using default test values...")
        test_rewards = [-1.0, -1.0, -1.0, 10.0]
        test_gamma = 0.9
        u_ret, d_ret = calculate_returns(test_rewards, test_gamma)
        print(f"Default Rewards: {test_rewards}")
        print(f"Default Gamma: {test_gamma}")
        print(f"Undiscounted Return: {u_ret:.4f}")
        print(f"Discounted Return: {d_ret:.4f}\n")


def run_assignment_2():
    print("=" * 50)
    print("ASSIGNMENT 2: 2x3 Grid World MDP Environment")
    print("=" * 50)

    # Instantiate environment
    env = GridWorldEnv()

    # a. Define State Space
    print(f"a. State Space (S): {env.get_state_space()}")

    # b. Define Action Space
    print(f"b. Action Space (A): {env.get_action_space()}")

    # d. Define Discount Factor
    print(f"d. Discount Factor (gamma): {env.get_discount_factor()}\n")

    # f. Test Transition Function with sample steps
    print("f. Transition Function Estimations (Sample Calculations):")
    sample_tests = [
        ("S0", "Right"),
        ("S0", "Down"),
        ("S1", "Right"),
        ("S3", "Left"),
        ("S4", "Up"),
    ]

    for curr_state, action in sample_tests:
        next_state, reward, is_terminal = env.transition_function(curr_state, action)
        print(f"Current State: {curr_state:<2} | Action: {action:<5} -> Next State: {next_state:<2} | Reward: {reward:<5.1f} | Episode Terminated: {is_terminal}")


if __name__ == "__main__":
    run_assignment_1()
    run_assignment_2()
```
