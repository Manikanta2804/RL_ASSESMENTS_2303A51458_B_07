# Reinforcement Learning: Lab Assignment 11
## Policy Iteration in a 2x2 Grid World

```python
import numpy as np

# 1. Define the MDP Parameters
states = ['S0', 'S1', 'G']
non_terminal_states = ['S0', 'S1']
actions = ['Up', 'Down', 'Left', 'Right']
gamma = 0.9
theta = 1e-6

# Transition and reward rules for the 2x2 grid environment
# Deterministic grid layout assumption based on output S0 -> S1 -> G:
# S0 | S1
# --------- (Wall/Boundary Below)
# G is reached from S1 via 'Right' (or another layout yielding G from S1)
# Let's map transition rules to match: S0 (Down->S0, Left->S0, Up->S0, Right->S1)
# S1 (Left->S0, Up->S1, Down->S1, Right->G)
def step(state, action):
    if state == 'G':
        return 'G', 0

    if state == 'S0':
        if action == 'Right':
            return 'S1', -1
        else: # Up, Down, Left hit walls
            return 'S0', -1

    if state == 'S1':
        if action == 'Right':
            return 'G', 100
        elif action == 'Left':
            return 'S0', -1
        else: # Up, Down hit walls
            return 'S1', -1

# Display Environment Details
print("=== MDP Configuration ===")
print(f"States: {states}")
print(f"Actions: {actions}")
print(f"Discount Factor (gamma): {gamma}")
```

```python
# 2. Policy Iteration Implementation

# Initialize state values to zero
V = {s: 0.0 for s in states}

# Initialize policy: S0 -> Up, S1 -> Up
policy = {'S0': 'Up', 'S1': 'Up'}

# Helper to evaluate action values (Q-values)
def compute_q_value(state, action, V_curr):
    next_state, reward = step(state, action)
    return reward + gamma * V_curr[next_state]

iteration = 1
policy_stable = False

print("=== Starting Policy Iteration ===\n")

while not policy_stable:
    print(f"--- Iteration {iteration} ---")

    # Step A: Policy Evaluation
    eval_step = 0
    while True:
        delta = 0
        V_new = V.copy()
        for s in non_terminal_states:
            v = V[s]
            chosen_action = policy[s]
            V_new[s] = compute_q_value(s, chosen_action, V)
            delta = max(delta, abs(v - V_new[s]))
        V = V_new
        eval_step += 1
        if delta < theta:
            break

    print(f"Policy Evaluation completed in {eval_step} steps.")
    print("State Values V(s):")
    for s in non_terminal_states:
        print(f"  V({s}) = {V[s]:.4f}")

    # Step B: Policy Improvement
    policy_stable = True
    print("\nCalculating Q-values for Policy Improvement:")
    for s in non_terminal_states:
        old_action = policy[s]
        q_values = {}
        for a in actions:
            q_values[a] = compute_q_value(s, a, V)

        # Select action with max Q-value
        best_action = max(q_values, key=q_values.get)
        policy[s] = best_action

        # Print action-value table for this state
        print(f"  State {s}: " + ", ".join([f"Q({a})={q_values[a]:.2f}" for a in actions]))

        if old_action != best_action:
            policy_stable = False

    print(f"Updated Policy: {{'S0': '{policy['S0']}', 'S1': '{policy['S1']}'}}")
    print("=" * 40 + "\n")
    iteration += 1

# =====================================================================
# 3. Final Results and Bellman Optimality Verification
# =====================================================================
print("=== FINAL OPTIMAL RESULTS ===")
print(f"{'State':<8}{'Optimal Value (V*)':<20}{'Optimal Action (pi*)'}")
for s in states:
    if s == 'G':
        print(f"{s:<8}{V[s]:<20.1f}{'Terminal'}")
    else:
        print(f"{s:<8}{V[s]:<20.1f}{policy[s]}")

print("\n=== Bellman Optimality Verification ===")
for s in non_terminal_states:
    q_vals = {a: compute_q_value(s, a, V) for a in actions}
    max_q = max(q_vals.values())
    verified = np.isclose(V[s], max_q)
    print(f"State {s}: V*(s) = {V[s]:.2f} | max_a Q*(s, a) = {max_q:.2f} | Matches? {verified}")
```
