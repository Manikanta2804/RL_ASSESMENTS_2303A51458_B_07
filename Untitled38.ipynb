```python
import numpy as np

# --- Parameters and Environment Setup ---
gamma = 0.9
theta = 1e-8  # Convergence threshold for policy evaluation

# States
# S0 = 0, S1 = 1, G (terminal) with V(G) = 0

# Policy
# pi[state][action] = probability
pi = {
    'S0': {'Down': 0.9, 'Right': 0.1},
    'S1': {'Right': 0.9, 'Up': 0.1}
}

# Transition Model and Rewards
# transitions[state][action] = [(probability, next_state, reward)]
# Note: 'Wall' stays in the same state according to Grid World conventions,
# but let's look at the assignment prompt's specific model:
# State S0 Down: Wall (0.1), S1 (0.9) | Reward: -1 for either outcome
# State S0 Right: Wall (1.0) | Reward: -1
# State S1 Right: G (0.9), S0 (0.1) | Reward: +100 to G, -1 to S0
# State S1 Up: S0 (1.0) | Reward: -1
#
# Let's model "Wall" as staying in the original state.
transitions = {
    'S0': {
        'Down':  [(0.1, 'S0', -1), (0.9, 'S1', -1)],
        'Right': [(1.0, 'S0', -1)]
    },
    'S1': {
        'Right': [(0.9, 'G', 100), (0.1, 'S0', -1)],
        'Up':    [(1.0, 'S0', -1)]
    }
}

# Initialize State Values
V = {'S0': 0.0, 'S1': 0.0, 'G': 0.0}

# --- Iterative Policy Evaluation ---
iteration = 0
while True:
    delta = 0
    # Store old values to compute delta
    v_old = V.copy()

    for s in ['S0', 'S1']:
        new_v = 0.0
        # Sum over actions: pi(a|s) * Q(s, a)
        for a, action_prob in pi[s].items():
            # Q(s, a) = sum_{s', r} P(s', r | s, a) * [r + gamma * V(s')]
            q_sa = 0.0
            for prob, next_state, r in transitions[s][a]:
                q_sa += prob * (r + gamma * V[next_state])
            new_v += action_prob * q_sa

        V[s] = new_v
        delta = max(delta, abs(v_old[s] - V[s]))

    iteration += 1
    if delta < theta:
        break

# --- Compute Action Values (Q-values) ---
Q = {}
for s in ['S0', 'S1']:
    Q[s] = {}
    for a in transitions[s].keys():
        q_sa = 0.0
        for prob, next_state, r in transitions[s][a]:
            q_sa += prob * (r + gamma * V[next_state])
        Q[s][a] = q_sa

# --- Print Results ---
print("=================== LAB ASSIGNMENT RESULTS ===================")
print(f"Number of evaluation iterations: {iteration}")
print(f"V_pi(S0) : {V['S0']:.6f}")
print(f"V_pi(S1) : {V['S1']:.6f}")
print(f"Q_pi(S0, Down) : {Q['S0']['Down']:.6f}")
print(f"Q_pi(S0, Right): {Q['S0']['Right']:.6f}")
print(f"Q_pi(S1, Right): {Q['S1']['Right']:.6f}")
print(f"Q_pi(S1, Up)   : {Q['S1']['Up']:.6f}")
print("==============================================================")

# --- Policy Consistency Checks ---
print("\n--- Verification & Consistency Checks ---")
calculated_V_S0 = pi['S0']['Down'] * Q['S0']['Down'] + pi['S0']['Right'] * Q['S0']['Right']
calculated_V_S1 = pi['S1']['Right'] * Q['S1']['Right'] + pi['S1']['Up'] * Q['S1']['Up']

print(f"Is V_pi(S0) equal to calculated sum? {np.isclose(V['S0'], calculated_V_S0)}")
print(f"Is V_pi(S1) equal to calculated sum? {np.isclose(V['S1'], calculated_V_S1)}")
```
