```python
# Multi-Armed Bandit - Greedy Algorithm
# Arithmetic Mean Method

# Take number of machines
n = int(input("Enter the number of machines: "))

machines = {}

# Input rewards for each machine
for i in range(n):
    name = input(f"\nEnter name of Machine {i + 1}: ")
    rewards = list(map(float, input(
        f"Enter observed rewards for Machine {name} (space-separated): "
    ).split()))

    machines[name] = rewards

print("\n--- Estimated Action Values ---")

q_values = {}

# Calculate Q(a) using arithmetic mean
for machine, rewards in machines.items():

    if len(rewards) > 0:
        q = sum(rewards) / len(rewards)
    else:
        q = 0

    q_values[machine] = q

    print(f"\nMachine {machine}")
    print(f"Rewards: {rewards}")
    print(f"Q({machine}) = {q:.2f}")

# Greedy selection
selected_machine = max(q_values, key=q_values.get)

print("\nGreedy Selected Machine:", selected_machine)
print(f"Highest Q value = {q_values[selected_machine]:.2f}")
```

```python
# Multi-Armed Bandit - Greedy Algorithm
# Incremental Update Method

# Take number of machines
n = int(input("Enter the number of machines: "))

# Initialize Q(a) and N(a)
Q = {}
N = {}

# Input machine names and rewards
machines = {}

for i in range(n):
    name = input(f"\nEnter name of Machine {i + 1}: ")

    rewards = list(map(float, input(
        f"Enter observed rewards for Machine {name} (space-separated): "
    ).split()))

    machines[name] = rewards

    # Initialize
    Q[name] = 0.0
    N[name] = 0

print("\n--- Incremental Updates ---")

# Process every observed reward
for machine, rewards in machines.items():

    for R in rewards:

        # 1. Increment number of plays
        N[machine] += 1

        # 2. Incremental update
        Q[machine] = Q[machine] + (
            (R - Q[machine]) / N[machine]
        )

        # Display update
        print(
            f"Machine {machine}: "
            f"Reward = {R:g}, "
            f"N = {N[machine]}, "
            f"Q = {Q[machine]:.2f}"
        )

# Display final Q values
print("\n--- Final Estimated Values ---")

for machine in Q:
    print(f"Q({machine}) = {Q[machine]:.2f}")

# Greedy selection
selected_machine = max(Q, key=Q.get)

print("\nGreedy Selected Machine:", selected_machine)
print(f"Highest Q value = {Q[selected_machine]:.2f}")
```
