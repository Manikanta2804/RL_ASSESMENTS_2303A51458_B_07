## Multi-Armed Bandit Problem: Upper Confidence Bound (UCB) Implementation

This notebook implements the UCB algorithm for a Multi-Armed Bandit problem with three slot machines (A, B, C) over 10 plays. The goal is to maximize cumulative rewards by balancing exploration and exploitation.

### Part A: Machine Initialization

We will initialize the machine names, the count of times each machine has been played (`N`), the total reward obtained from each machine (`R`), the estimated value (Q-value) of each machine (`Q`), and the Upper Confidence Bound (`UCB`) for each machine. We also define the total number of plays.

```python
import numpy as np

# Define the machines
machines = ['A', 'B', 'C']
M = len(machines) # Number of machines
N_play = 10 # Total number of plays

# Initialize data structures for each machine
# N: Number of times each machine has been played
# R: Total reward obtained from each machine
# Q: Estimated value (Q-value) of each machine
# UCB: Upper Confidence Bound for each machine
N = {machine: 0 for machine in machines}
R = {machine: 0 for machine in machines}
Q = {machine: 0.0 for machine in machines} # Q-values as float
UCB = {machine: 0.0 for machine in machines}

total_plays = 0 # To keep track of the total number of plays made
total_reward_player = 0 # To accumulate the total reward received by the player

# Reward patterns for simulation (as per problem statement)
reward_patterns = {
    'A': [3, 4, 2, 5],
    'B': [6, 7, 5, 8],
    'C': [4, 3, 5, 4]
}

# Keep track of the current index for each machine's reward pattern
reward_indices = {machine: 0 for machine in machines}

print("Initialization complete:")
print(f"Machines: {machines}")
print(f"Initial N: {N}")
print(f"Initial R: {R}")
print(f"Initial Q: {Q}")
print(f"Initial UCB: {UCB}")
```

### Parts C, D, E: UCB Calculation, Machine Selection, Reward Collection, and Q-Value Update

Now, we will run the simulation for `N_play` times. In each play:
1.  **Calculate UCB** for all machines. We'll use a exploration constant `c=2` for the UCB formula: `UCB(a) = Q(a) + c * sqrt(log(total_plays) / N(a))`. If a machine hasn't been played yet (`N(a)=0`), its UCB will be set to a very high value to ensure it gets explored.
2.  **Choose the machine** with the highest UCB value.
3.  **Simulate Reward Collection** using the predefined `reward_patterns`. We will cycle through the provided rewards for each machine.
4.  **Update `N` and `R`** for the selected machine.
5.  **Update `Q`-value** using the incremental sample-average formula: `Q(a) = Q(a) + (reward - Q(a)) / N(a)`.

```python
c = 2 # Exploration constant for UCB

for play_num in range(1, N_play + 1):
    print(f"\n--- Play {play_num} ---")
    total_plays += 1

    # 1. Calculate UCB for each machine
    for machine in machines:
        if N[machine] == 0:
            UCB[machine] = float('inf') # Prioritize unplayed machines
        else:
            # UCB formula: Q(a) + c * sqrt(log(total_plays) / N(a))
            UCB[machine] = Q[machine] + c * np.sqrt(np.log(total_plays) / N[machine])

    print(f"UCB values before selection: {UCB}")

    # 2. Choose machine that has highest UCB value to play
    selected_machine = max(UCB, key=UCB.get)
    print(f"Selected machine for play {play_num}: {selected_machine}")

    # 3. Simulate reward obtained from the selected machine
    # Use the predefined reward pattern and cycle through it
    current_reward_index = reward_indices[selected_machine]
    reward = reward_patterns[selected_machine][current_reward_index]
    reward_indices[selected_machine] = (current_reward_index + 1) % len(reward_patterns[selected_machine])

    print(f"Reward obtained from machine {selected_machine}: {reward}")
    total_reward_player += reward

    # 4. Update the number of times the selected machine has been played
    N[selected_machine] += 1
    R[selected_machine] += reward # Update total reward for this machine

    # 5. Update its Q-value using the incremental sample-average formula
    # Q(a) = Q(a) + (R - Q(a)) / N(a)
    Q[selected_machine] = Q[selected_machine] + (reward - Q[selected_machine]) / N[selected_machine]

    print(f"Updated N: {N}")
    print(f"Updated R: {R}")
    print(f"Updated Q: {Q}")
```

### Part F: Display Results

After all 10 plays, let's display the final state of our knowledge about the machines, including their play counts (`N`), total rewards (`R`), estimated Q-values (`Q`), and the last calculated UCB values. We will also display the total reward accumulated by the player throughout the simulation.

```python
print("\n--- Final Results ---")
print(f"Total plays made: {total_plays}")
print(f"Final N (Plays per machine): {N}")
print(f"Final R (Total reward per machine): {R}")
print(f"Final Q (Estimated value per machine): {Q}")

# Recalculate UCB values one last time based on final N, Q, and total_plays
final_ucb = {}
for machine in machines:
    if N[machine] == 0:
        final_ucb[machine] = float('nan') # Not played, UCB is undefined
    else:
        final_ucb[machine] = Q[machine] + c * np.sqrt(np.log(total_plays) / N[machine])
print(f"Final UCB values: {final_ucb}")

print(f"\nTotal reward received by the player: {total_reward_player}")
```
