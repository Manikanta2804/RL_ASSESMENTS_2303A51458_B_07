## Part A – Machine Initialization and Problem Setup

```python
import random

# Problem parameters
M = 3  # Number of machines
machines = ['A', 'B', 'C']
N_play = 10  # Number of plays
epsilon = 0.2  # Exploration probability (ε)

# Initialize N (play count), R (total reward), and Q (estimated value) for each machine
# Using dictionaries for easy access by machine name
N = {machine: 0 for machine in machines}  # Number of times each machine has been played
R = {machine: 0.0 for machine in machines} # Total reward obtained from each machine
Q = {machine: 0.0 for machine in machines} # Estimated Q-value (average reward) for each machine

total_reward_player = 0.0 # To track the total reward received by the player

print("Initialization complete:")
print(f"Machines: {machines}")
print(f"Initial N: {N}")
print(f"Initial R: {R}")
print(f"Initial Q: {Q}")
print(f"Total Plays: {N_play}, Exploration Probability (ε): {epsilon}")
```

## Multi-Armed Bandit Simulation

This section implements the modified ε-greedy strategy, reward collection, and Q-value updates for each play.

```python
for play_num in range(1, N_play + 1):
    print(f"\n--- Play {play_num} ---")
    selected_machine = None
    r = random.random()

    if r < epsilon: # Exploration (Part C - Modified ε-Greedy Decision)
        print("Performing exploration...")
        # Find the machine(s) with the highest current Q-value
        max_q_value = max(Q.values())
        machines_with_max_q = [machine for machine, q_val in Q.items() if q_val == max_q_value]

        if len(machines_with_max_q) == M: # All Q-values are equal (e.g., at the start)
            # If all machines have the same Q-value, explore any machine randomly
            selected_machine = random.choice(machines)
        else:
            # Select one of the machines EXCEPT the one with the highest estimated Q-value
            non_highest_q_machines = [machine for machine in machines if machine not in machines_with_max_q]
            selected_machine = random.choice(non_highest_q_machines)
    else: # Exploitation (Part C - Modified ε-Greedy Decision)
        print("Performing exploitation...")
        # Select the machine having the highest current Q-value
        max_q_value = max(Q.values())
        # If multiple machines have the same max Q-value, pick one randomly among them
        machines_to_exploit = [machine for machine, q_val in Q.items() if q_val == max_q_value]
        selected_machine = random.choice(machines_to_exploit)

    print(f"Selected machine: {selected_machine}")

    # Part D – Reward Collection
    while True:
        try:
            reward_input = input(f"Enter reward obtained from machine {selected_machine}: ")
            reward = float(reward_input)
            break
        except ValueError:
            print("Invalid input. Please enter a numerical reward.")

    total_reward_player += reward

    # Part E – Q-Value Update
    N[selected_machine] += 1
    R[selected_machine] += reward

    # Incremental sample-average formula
    # Q(a) = Q(a) + (Reward - Q(a)) / N(a)
    Q[selected_machine] = Q[selected_machine] + (reward - Q[selected_machine]) / N[selected_machine]

    print(f"Reward received: {reward}")
    print(f"Updated N for {selected_machine}: {N[selected_machine]}")
    print(f"Updated R for {selected_machine}: {R[selected_machine]:.2f}")
    print(f"Updated Q for {selected_machine}: {Q[selected_machine]:.2f}")
```

## Part F – Display Results

Final participation counts, total rewards, and estimated Q-values for each machine, along with the total reward accumulated by the player.

```python
print("\n=== Final Results ===")
print("\nParticipation Count (N):")
for machine, count in N.items():
    print(f"  Machine {machine}: {count} plays")

print("\nTotal Rewards (R):")
for machine, total_r in R.items():
    print(f"  Machine {machine}: {total_r:.2f}")

print("\nEstimated Q-Values (Q):")
for machine, q_val in Q.items():
    print(f"  Machine {machine}: {q_val:.2f}")

print(f"\nTotal reward received by the player across all {N_play} plays: {total_reward_player:.2f}")
```
