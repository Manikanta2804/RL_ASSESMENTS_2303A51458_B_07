```python
# grid_env_det.py

# State Space
states = ["S0", "S1", "S2", "S3", "S4", "S5"]

# Action Space
actions = ["up", "down", "left", "right"]

# Discount Factor
gamma = 0.9

# Reward Function
# Goal state = S5
# Normal movement reward = -1
# Goal reward = +10
rewards = {
    "S0": -1,
    "S1": -1,
    "S2": -1,
    "S3": -1,
    "S4": -1,
    "S5": 10
}


# Transition Function
def transition(state, action):

    transitions = {
        "S0": {
            "up": "S0",
            "down": "S3",
            "left": "S0",
            "right": "S1"
        },

        "S1": {
            "up": "S1",
            "down": "S1",
            "left": "S0",
            "right": "S2"
        },

        "S2": {
            "up": "S2",
            "down": "S5",
            "left": "S1",
            "right": "S2"
        },

        "S3": {
            "up": "S0",
            "down": "S3",
            "left": "S3",
            "right": "S4"
        },

        "S4": {
            "up": "S1",
            "down": "S4",
            "left": "S3",
            "right": "S5"
        },

        "S5": {
            "up": "S5",
            "down": "S5",
            "left": "S5",
            "right": "S5"
        }
    }

    return transitions[state][action]
```

```python
# Practical - 8
# Calculate State Value and Action Value
# Without using Bellman Equation

# Policy updated to prevent infinite loops and reach S5
policy = {
    "S0": "right",
    "S1": "right",
    "S2": "down",
    "S3": "right",
    "S4": "right"
}


# Discount factor
# gamma is already imported in the environment cell


# -------------------------------------------------
# Calculate State Value
# -------------------------------------------------

def state_value_function(start_state):

    current_state = start_state
    G = 0
    discount = 1

    print("\nState Value Calculation")
    print("-----------------------")

    while current_state != "S5":

        action = policy[current_state]

        next_state = transition(
            current_state,
            action
        )

        reward = rewards[next_state]

        print(
            current_state,
            "--", action, "-->",
            next_state,
            "Reward =", reward
        )

        G = G + discount * reward

        discount = discount * gamma

        current_state = next_state

    return G


# -------------------------------------------------
# Calculate Action Value
# -------------------------------------------------

def action_value_function(start_state, action):

    current_state = start_state

    # First take the given action
    next_state = transition(
         current_state,
         action
    )

    reward = rewards[next_state]

    G = reward

    discount = gamma

    print("\nAction Value Calculation")
    print("-----------------------")

    print(
        current_state,
        "--", action, "-->",
        next_state,
        "Reward =", reward
    )

    current_state = next_state

    # Continue following the policy
    while current_state != "S5":

        action = policy[current_state]

        next_state = transition(
            current_state,
            action
        )

        reward = rewards[next_state]

        print(
            current_state,
            "--", action, "-->",
            next_state,
            "Reward =", reward
        )

        G = G + discount * reward

        discount = discount * gamma

        current_state = next_state

    return G


# -------------------------------------------------
# Main Program
# -------------------------------------------------

print("REINFORCEMENT LEARNING - PRACTICAL 8")
print("------------------------------------")

# State Value of S0
V_S0 = state_value_function("S0")

print("\nState Value of S0 =", V_S0)


# Action Value of S0 for action UP
Q_S0_UP = action_value_function("S0", "up")

print("\nAction Value of S0 for action UP =", Q_S0_UP)
```
