```python
import numpy as np
import pandas as pd

# 1. Define the MDP environment parameters
states = ['S0', 'S1', 'G']
non_terminal_states = ['S0', 'S1']
actions = ['Up', 'Down', 'Left', 'Right']
gamma = 0.9
theta = 1e-6 # small threshold for convergence checking

# Define transitions and rewards dynamically according to the problem
# G is terminal.
# S0 actions: Down -> S1 (reward -1), Up/Left/Right -> S0 (reward -1)
# S1 actions: Right -> G (reward +100), Up -> S0 (reward -1), Down/Left -> S1 (reward -1)
def step(state, action):
    if state == 'G':
        return 'G', 0

    if state == 'S0':
        if action == 'Down':
            return 'S1', -1
        else: # Up, Left, Right hit the boundary/wall
            return 'S0', -1

    if state == 'S1':
        if action == 'Right':
            return 'G', 100
        elif action == 'Up':
            return 'S0', -1
        else: # Down, Left hit the boundary/wall
            return 'S1', -1

# Print MDP Environment Information
print("=== MDP ENVIRONMENT INFORMATION ===")
transitions_info = []
for s in non_terminal_states:
    for a in actions:
        next_s, r = step(s, a)
        transitions_info.append({"Current State": s, "Action": a, "Next State": next_s, "Reward": r})
df_mdp = pd.DataFrame(transitions_info)
print(df_mdp.to_string(index=False))
print("\n" + "="*40 + "\n")

# 2. Initialize parameters
V = {s: 0.0 for s in states} # Initialize state values to zero
iteration_history = []

# Keep track of initial iteration state values
iteration_history.append({"Iteration": 0, "V(S0)": V['S0'], "V(S1)": V['S1'], "V(G)": V['G']})

# Value Iteration Loop
iteration = 1
converged = False

while not converged:
    delta = 0
    next_V = V.copy()

    for s in non_terminal_states:
        best_value = -float('inf')
        for a in actions:
            next_s, r = step(s, a)
            q_value = r + gamma * V[next_s]
            if q_value > best_value:
                best_value = q_value

        delta = max(delta, abs(best_value - V[s]))
        next_V[s] = best_value

    V = next_V
    iteration_history.append({"Iteration": iteration, "V(S0)": V['S0'], "V(S1)": V['S1'], "V(G)": V['G']})

    if delta < theta:
        converged = True
    iteration += 1

# Display Iteration-wise State Value Table
print("=== ITERATION-WISE STATE-VALUE TABLE ===")
df_history = pd.DataFrame(iteration_history)
print(df_history.to_string(index=False))
print("\n" + "="*40 + "\n")

# Calculate final optimal action-value Q*(s, a) table & optimal policy
q_table = []
optimal_policy = {}

for s in non_terminal_states:
    best_action = None
    best_q = -float('inf')
    for a in actions:
        next_s, r = step(s, a)
        q_val = r + gamma * V[next_s]
        q_table.append({"State": s, "Action": a, "Next State": next_s, "Q*(s,a)": q_val})
        if q_val > best_q:
            best_q = q_val
            best_action = a
    optimal_policy[s] = (best_action, best_q)

# Display Final Action-Value Table
print("=== FINAL ACTION-VALUE TABLE Q*(s, a) ===")
df_q = pd.DataFrame(q_table)
print(df_q.to_string(index=False))
print("\n" + "="*40 + "\n")

# Display Optimal Policies and Final State Values
print("=== OPTIMAL STATE VALUES AND OPTIMAL POLICY ===")
for s in non_terminal_states:
    opt_action, opt_val = optimal_policy[s]
    print(f"State: {s} | Optimal Value V*({s}): {opt_val:.2f} | Optimal Action: {opt_action}")
print(f"State: G  | Optimal Value V*(G) : {V['G']:.2f}  | (Terminal State)")
```
