### Implement a Random Agent for Reinforcement Learning Environments

This section will implement a `RandomAgent` class that interacts with different Gym environments by selecting actions randomly. We will test this agent on the `CliffWalking-v0` and `Taxi-v3` environments to observe basic agent-environment interaction and reward calculation.

```python
import gymnasium as gym
import random
import numpy as np

print("Libraries imported successfully!")
```

### RandomAgent Class Definition

The `RandomAgent` class encapsulates the logic for creating an environment, interacting with it by choosing random actions, and tracking rewards over multiple episodes.

```python
class RandomAgent:
    def __init__(self, env_name):
        self.env = gym.make(env_name)
        self.env_name = env_name
        self.action_space_size = self.env.action_space.n
        print(f"Initialized RandomAgent for environment: {self.env_name}")
        print(f"Action space size: {self.action_space_size}")

    def run_episode(self, episode_idx):
        state, info = self.env.reset()
        done = False
        total_reward = 0
        step = 0

        while not done:
            action = self.env.action_space.sample() # Select a random action
            next_state, reward, terminated, truncated, info = self.env.step(action)
            done = terminated or truncated
            total_reward += reward
            state = next_state
            step += 1

        print(f"Episode {episode_idx + 1}: Total Reward = {total_reward:.2f}, Steps = {step}")
        return total_reward

    def train(self, num_episodes):
        episode_rewards = []
        print(f"\n--- Starting training for {self.env_name} with {num_episodes} episodes ---")
        for i in range(num_episodes):
            reward = self.run_episode(i)
            episode_rewards.append(reward)

        self.env.close()

        average_reward = np.mean(episode_rewards)
        print(f"\n--- Training complete for {self.env_name} ---")
        print(f"Total rewards across {num_episodes} episodes: {np.sum(episode_rewards):.2f}")
        print(f"Average reward per episode: {average_reward:.2f}")
        return episode_rewards
```

### Test with CliffWalking-v0 Environment

Now, let's create an instance of the `RandomAgent` and run it on the `CliffWalking-v0` environment for a few episodes.

```python
cliff_walking_agent = RandomAgent('CliffWalking-v1')
cliff_walking_rewards = cliff_walking_agent.train(num_episodes=10)
```

### Test with Taxi-v3 Environment

Next, we'll test the `RandomAgent` on the `Taxi-v3` environment.

```python
taxi_agent = RandomAgent('Taxi-v4')
taxi_rewards = taxi_agent.train(num_episodes=10)
```
