## Objective 1: Greedy Algorithm with Arithmetic Mean

```python
# Observed rewards for each machine
rewards = {
    'A': [3, 2, 4],
    'B': [6, 7],
    'C': [5, 4, 6]
}

# Calculate Q(a) for each machine using the arithmetic mean
q_values = {}
for machine, reward_list in rewards.items():
    q_values[machine] = sum(reward_list) / len(reward_list)

# Display the estimated Q(a) for every machine
for machine, reward_list in rewards.items():
    print(f"Machine {machine}")
    print(f"Rewards: {reward_list}")
    print(f"Q({machine}) = {q_values[machine]:.2f}")

# Select the machine with the highest Q(a) (Greedy approach)
greedy_selected_machine = max(q_values, key=q_values.get)

# Display the selected machine
print(f"\nGreedy Selected Machine: {greedy_selected_machine}")
```

## Objective 2: Greedy Algorithm with Incremental Update Method

```python
# Initial values as per Task 1 – Initialize
q_values = {
    'A': 0.0,
    'B': 0.0,
    'C': 0.0
}
n_values = {
    'A': 0,
    'B': 0,
    'C': 0
}

# Observed rewards for each machine (as per problem description)
observed_rewards = {
    'A': [3, 2, 4],
    'B': [6, 7],
    'C': [5, 4, 6]
}

print("Initial State:")
for machine in q_values:
    print(f"Q({machine}) = {q_values[machine]:.2f}, N({machine}) = {n_values[machine]}")
print("\n")

# Simulate plays and update Q(a) incrementally
# For simplicity, we'll process rewards machine by machine as they are given.
# In a true MAB scenario, the agent would choose an arm at each step.
# Here, we assume the 'observed rewards' are the sequence of plays.

all_plays = []
for machine, rewards_list in observed_rewards.items():
    for reward in rewards_list:
        all_plays.append({'machine': machine, 'reward': reward})

# Sort plays by machine for consistent output, or process as a stream if order matters
# For this problem, processing the given reward lists as they appear is sufficient for demonstration

play_counter = 0
for machine_to_play in ['A', 'B', 'C']:
    print(f"Processing rewards for Machine {machine_to_play}:")
    for reward in observed_rewards[machine_to_play]:
        play_counter += 1
        print(f"  --- Play {play_counter} ---")
        selected_machine = machine_to_play # The current problem statement implies we're processing observed rewards per machine

        # 3. Increment N(a)
        n_values[selected_machine] += 1
        n_a = n_values[selected_machine]

        # 4. Update Q(a) using the incremental formula
        q_old_a = q_values[selected_machine]
        q_new_a = q_old_a + (reward - q_old_a) / n_a
        q_values[selected_machine] = q_new_a

        print(f"  Selected Machine: {selected_machine}, Observed Reward: {reward}")
        for m in q_values:
            print(f"  Q({m}) = {q_values[m]:.2f}, N({m}) = {n_values[m]}")

        # 6. Identify the machine having the highest current Q(a)
        highest_q_machine = max(q_values, key=q_values.get)
        print(f"  Current Highest Q Machine: {highest_q_machine}\n")

print("Final Estimated Q(a) values after all observed rewards:")
for machine in q_values:
    print(f"Q({machine}) = {q_values[machine]:.2f}, N({machine}) = {n_values[machine]}")

final_greedy_selected_machine = max(q_values, key=q_values.get)
print(f"\nFinal Greedy Selected Machine (after all plays): {final_greedy_selected_machine}")
```
