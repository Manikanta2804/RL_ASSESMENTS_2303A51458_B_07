```python
import random

machines = ['A', 'B', 'C']

N_play = 10
epsilon = 0.2

# Predefined reward patterns for the suggested experiment
# These lists represent the rewards to be given in order for each machine.
reward_patterns = {
    'A': [3, 4, 2, 5],
    'B': [6, 7, 5, 8],
    'C': [4, 3, 5, 4]
}

N = {
    'A': 0,
    'B': 0,
    'C': 0
}

R = {
    'A': [],
    'B': [],
    'C': []
}

Q = {
    'A': 0.0,
    'B': 0.0,
    'C': 0.0
}

print("==============================================")
print("   ε-GREEDY MULTI-ARMED BANDIT")
print("==============================================")

print("\nMachines:", machines)
print("Number of plays:", N_play)
print("Epsilon:", epsilon)

print("\nInitial Tables:")
print("N =", N)
print("R =", R)
print("Q =", Q)

for play in range(1, N_play + 1):

    r = random.random()

    print("\n----------------------------------------------")
    print("Play", play)
    print("Random number r =", round(r, 4))

    if r < epsilon:
        selected_machine = random.choice(machines)

        print("Decision: EXPLORATION")
        print("Randomly selected machine:", selected_machine)

    else:
        # In case all Q values are 0.0 initially, max(Q.values()) will be 0.0
        # and all machines will be in best_machines. random.choice will pick one.
        max_Q = max(Q.values())

        best_machines = [
            machine for machine in machines
            if Q[machine] == max_Q
        ]

        selected_machine = random.choice(best_machines)

        print("Decision: EXPLOITATION")
        print("Highest Q-value:", max_Q)
        print("Selected machine:", selected_machine)

    print("\nPlay with machine", selected_machine)

    # Retrieve reward from the predefined pattern based on the number of times this machine has been played
    # N[selected_machine] tracks how many times it has been played *before* this turn
    # so it serves as the correct index for the next reward in the pattern.
    try:
        reward = float(reward_patterns[selected_machine][N[selected_machine]])
    except IndexError:
        print(f"Warning: Not enough predefined rewards for machine {selected_machine}. Assigning 0.0.")
        reward = 0.0 # Assign a default reward if pattern runs out

    R[selected_machine].append(reward)

    N[selected_machine] += 1

    old_Q = Q[selected_machine]

    Q[selected_machine] = (
        old_Q + (reward - old_Q) / N[selected_machine]
    )

    print("\nUpdated values:")
    print("Machine:", selected_machine)
    print("Reward:", reward)
    print("N =", N[selected_machine])
    print("Old Q =", round(old_Q, 4))
    print("New Q =", round(Q[selected_machine], 4))

print("\n\n==============================================")
print("             FINAL RESULTS")
print("==============================================")

print("\nN TABLE - Participation Count")
print("----------------------------------------------")

for machine in machines:
    print(f"Machine {machine}: N({machine}) = {N[machine]}")

print("\nR TABLE - Observed Rewards")
print("----------------------------------------------")

for machine in machines:
    print(f"Machine {machine}: R({machine}) = {R[machine]}")

print("\nQ TABLE - Estimated Values")
print("----------------------------------------------")

for machine in machines:
    print(f"Machine {machine}: Q({machine}) = {Q[machine]:.2f}")

best_machine = max(Q, key=Q.get)

print("\n==============================================")
print("Greedy Best Machine:", best_machine)
print("Highest Q-value:", round(Q[best_machine], 2))
print("==============================================")
```
